```python
# 取出台灣股市 ETF
df = api.taiwan_stock_info()
```

```python
Bond_dataframe = df[
    (df["stock_name"].str.contains("債"))
    & (df["stock_name"].str.contains("元大"))
]
Bond_dataframe
```

```python
ETF_dataframe = df[
    (df["industry_category"].str[:3] == "ETF")
    & (df["stock_name"].str.contains("元大"))
    & (df["stock_id"].str[-1].str.isnumeric())
]
ETF_dataframe
```
